### Nominal/OHE Encoding

One hot encoding, also known as nominal endocing, is a technique used to present categorical data as numerical data which is more suitable for machine learning algorithms.  In this technique, each category is represented as a binary vector where each bit corrosponds to a unique category. For example, if we have a categorical variable 'Color' with three possible (red,green,blue) we can represent it one hot encoding is follows: 

1. Red:[1,0,0]
2. Green:[0,1,0]
3. Blue:[0,0,1]
4. 

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
## Create a simple dataFrame

df=pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df.head()
```

```python
## Create an instance of Onehotencoder
encoder=OneHotEncoder()
```

```python
## Perform fit and transform 
enc=encoder.fit_transform(df[['color']]).toarray()
```

```python
import pandas as pd
```

```python
enc_df = pd.DataFrame(
    encoder.fit_transform(df[['color']]).toarray(),
    columns = encoder.get_feature_names_out()
)
```

```python
enc_df
```

```python
## For new data

encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,enc_df],axis=1)
```
