# Polynomial Regression

```python
#importing required library 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x=6*np.random.rand(100,1)-3
y=0.5* x**2 + 1.5* x + 2 + np.random.randn(100,1)
```

```python
len(x)
```

```python
len(y)
```

```python
plt.scatter(x,y,c='r')
plt.xlabel('x_data point')
plt.ylabel('y_data point ')
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
# Let's Implement simple linear regression 
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
```

```python
regression.fit(x_train,y_train)
```

```python
regression
```

```python
from sklearn.metrics import r2_score
```

```python
score=r2_score(y_test,regression.predict(x_test))
print(score)
```

```python
# Let's visualize this model 
plt.plot(x_train,regression.predict(x_train),color='g')
plt.scatter(x_train,y_train)
plt.xlabel('x_data point ')
plt.ylabel('y_data poinnt ')
```

```python
## Let's Apply polynomial Transformation 
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly=PolynomialFeatures(degree=2,include_bias=True)
```

```python
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
#x_train_poly
```

```python
from sklearn.metrics import r2_score
```

```python
regression=LinearRegression()
```

```python
regression.fit(x_train_poly,y_train)
```

```python
y_pred=regression.predict(x_test_poly)
```

```python
score=r2_score(y_test,y_pred)
print(score)
```

```python
regression.coef_
```

```python
regression.intercept_
```

```python
plt.scatter(x_train,regression.predict(x_train_poly))
plt.scatter(x_train,y_train)
```

```python
# Prediction of new data point 
x_new=np.linspace(-3,3,200).reshape(200,1)
x_new_poly=poly.transform(x_new)
```

```python
y_new=regression.predict(x_new_poly)
```

```python
plt.plot(x_new,y_new,'r',linewidth=2,label='predictions')
plt.plot(x_train,y_train,'b.',label='Traning point ')
plt.plot(x_test,y_test,'g.',label='Testing Point')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
