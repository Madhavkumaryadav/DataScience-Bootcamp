```python
## Independent and dependent features
x=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df['total_bill']
```

```python
x.head()
```

```python
y.head()
```

```python
## Train test split 
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
x_train
```

```python
## Feature Encoding(Label encoding and onehot Endocing ()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
x_train.head()
```

```python
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
x_train['sex']=le1.fit_transform(x_train['sex'])
x_train['smoker']=le2.fit_transform(x_train['smoker'])
x_train['time']=le3.fit_transform(x_train['time'])
```

```python
x_train.head()
```

```python
x_test['sex']=le1.fit_transform(x_test['sex'])
x_test['smoker']=le2.fit_transform(x_test['smoker'])
x_test['time']=le3.fit_transform(x_test['time'])
```

```python
x_test.head()
```

```python
## Apply the day columns in the Onehot encoding 

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first'), [3])
    ],
    remainder='passthrough'
)
```

```python
# ct.fit_transform(x_train)
```

```python
import sys
import numpy as np
np.set_printoptions(threshold=sys.maxsize)
x_train=ct.fit_transform(x_train)
```

```python
x_test=ct.fit_transform(x_test)
```

```python
x_train
```

```python
## Applying SVR -- SUpport vector Regression 
from sklearn.svm import SVR
```

```python
svr=SVR()
```

```python
svr.fit(x_train,y_train)
```

```python
y_pred=svr.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```

```python
## Hyperparameter Tuning Using Grid Search CV

from sklearn.model_selection import GridSearchCV

#defining parameters range
param_grid={'C':[0.1,1,10,100,1000],
            'gamma':[1,0.1,0.01,0.001,0.0001],
            'kernel':['rbf']}
```

```python
cv=GridSearchCV(SVR(),param_grid,refit=True,verbose=3)

#fitting the model for grid search 
cv.fit(x_train,y_train)
```

```python
cv.best_params_
```

```python
y_pred1=cv.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_pred1))
print(mean_absolute_error(y_test,y_pred1))
```
