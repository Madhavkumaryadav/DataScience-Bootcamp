```python
## FIrest remove the unnessarary columns 
data.drop('CustomerID',axis=1,inplace=True)
```

```python
cat_null=[]
num_null=[]
obj=[]
for i in data.columns:
    if data[i].dtypes=='object' and data[i].isnull().sum() != 0:
        cat_null.append(i)
    elif data[i].dtypes != 'object' and data[i].isnull().sum() !=0:
        num_null.append(i)
```

```python
cat_null
```

```python
num_null
```

```python
# for i in num_null:
#     data[i]=data[i].fillna(data[i].mean())

# for j in cat_null:
#     data[j]=data[j].fillna(data[j].mode()[0])

data.Age.fillna(data.Age.median(),inplace=True)

data.DurationOfPitch.fillna(data.DurationOfPitch.median(),inplace=True)
data.TypeofContact.fillna(data.TypeofContact.mode()[0],inplace=True)
data.NumberOfFollowups.fillna(data.NumberOfFollowups.mode()[0],inplace=True)
data.PreferredPropertyStar.fillna(data.PreferredPropertyStar.mode()[0],inplace=True)
data.NumberOfTrips.fillna(data.NumberOfTrips.median(),inplace=True)
data.NumberOfChildrenVisiting.fillna(data.NumberOfChildrenVisiting.mode()[0],inplace=True)
data.MonthlyIncome.fillna(data.MonthlyIncome.median(),inplace=True)
```

```python
data.isnull().sum()
```

```python
## Features Engineering 
## Create new column for feature
data['TotalVisiting']=data['NumberOfPersonVisiting'] + data['NumberOfChildrenVisiting']
data.drop(columns=['NumberOfPersonVisiting','NumberOfChildrenVisiting'])

data.head()
```

```python
data['Gender'].value_counts()
```

```python
data['Gender']=data['Gender'].replace('Fe Male','Female')
```

```python
data['Gender'].value_counts()
```

```python
data.isnull().sum()  
```

```python
cate=[]
for i in data.columns:
    if data[i].dtypes =='object':
        cate.append(i)
```

```python
from sklearn.preprocessing import LabelEncoder
l1=LabelEncoder()
```

```python
for i in cate:
    data[i]=l1.fit_transform(data[i])
```

```python
data.head()
data['Age']=[int(i) for i in data['Age']]
```

```python
data.head()
```

```python
x=data.drop('OwnCar',axis=1)
y=data['OwnCar']
```

```python
x.shape,y.shape
```

```python
## split the data in into train and test set 
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
x_train.shape,y_train.shape
```

```python
x_test.shape,y_test.shape
```

```python
## Train the model for xgboost 
from xgboost import XGBClassifier
```

```python
xgbr=XGBClassifier()
```

```python
xgbr.fit(x_train,y_train)
```

```python
## prediction 
from sklearn.metrics import confusion_matrix,classification_report
y_pred_test=xgbr.predict(x_test)
y_pred_train=xgbr.predict(x_train)

cm_train=confusion_matrix(y_pred_train,y_train)
cm_test=confusion_matrix(y_pred_test,y_test)
cr_train=classification_report(y_pred_train,y_train)
cr_test=classification_report(y_pred_test,y_test)

print('='*50 , 'Training Accuracy ' , '='*50)
print(cm_train)
print(cr_train)
print('='*50 , 'Testing Accuracy ' , '='*50)
print(cm_test)
print(cr_test)
```

```python
## Xgboost hyperparameter 
param_grid = {
    "objective": ["multi:softprob"],
    "num_class": [2, 3, 4, 5, 6],   # 1 valid nahi hota
    "eval_metric": ["mlogloss"],

    "n_estimators": [100, 200, 300],
    "learning_rate": [0.01, 0.05, 0.1],
    "max_depth": [3, 5, 7],
    "min_child_weight": [1, 3, 5],
    "gamma": [0, 0.1, 0.2],

    "subsample": [0.6, 0.8, 1.0],
    "colsample_bytree": [0.6, 0.8, 1.0],

    "reg_alpha": [0, 0.1, 1],
    "reg_lambda": [1, 1.5, 2]
}
```
