```python
## import requirement library 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
## Load the dataset library 
from sklearn.datasets import load_wine
```

```python
wine=load_wine()
```

```python
data=pd.DataFrame(wine.data)
```

```python
data.head()
```

```python
data.columns=wine.feature_names
```

```python
data.head()
```

```python
plt.scatter(data['hue'],data['proline'])
```

```python
from sklearn.ensemble import IsolationForest
d={'hue':data['hue'],
   'proline':data['proline']
  }
df=pd.DataFrame(data=d)
df.head()
```

```python
clf=IsolationForest(contamination=0.2)
clf.fit(df)
prediction=clf.predict(df)
```

```python
prediction
```

```python
## Index is the index of the outlier data point 

index=np.where(prediction < 0)
index
```

```python
x=df.values
```

```python
abn_ind=np.where(prediction < 0)
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[index,0],x[index,1],edgecolors='r')
```
