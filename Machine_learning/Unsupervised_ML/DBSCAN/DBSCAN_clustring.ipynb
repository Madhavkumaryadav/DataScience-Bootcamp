## DBASCAN Clustring

```python
from sklearn.cluster import DBSCAN 
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.datasets import make_circles
```

```python
x,y=make_circles(n_samples=750,factor=0.3,noise=0.1)
```

```python
plt.scatter(x[:,0],x[:,1],c=y)
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan=DBSCAN(eps=0.10)
```

```python
dbscan.fit(x)
```

```python
dbscan.fit_predict(x)
```

```python
dbscan.labels_
```

```python
plt.scatter(x[:,0],x[:,1],c=dbscan.labels_)
```

```python
plt.scatter(x[:,0],x[:,1],c=y)
```

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
lof=LocalOutlierFactor(n_neighbors=5,algorithm='auto',leaf_size=10,metric='minkowski',p=2)
```

```python
lof.fit(x)
```

```python
lof.get_params
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param={
    'n_neighbors':[i for i in range(1,51,5)],
    'alogrithm':['auto','ball_tree','kd_tree','brute'],
    'leaf_size':[i for i in range(10,50,4)],
    'metric':'minkowski',
    'p':[i for i in range(1,20,2)],
    
}
```
