## Principle Component analysis 

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


%matplotlib inline
```

```python
## Load dataset 
from sklearn.datasets import load_breast_cancer
```

```python
cancer_data=load_breast_cancer()
```

```python
cancer_data.keys()
```

```python
print(cancer_data.DESCR)
```

```python
df=pd.DataFrame(cancer_data['data'],columns=cancer_data['feature_names'])
```

```python
df.head()
```

```python
## Standarization 
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
scaler.fit(df)
```

```python
scaled_data=scaler.transform(df)
```

```python
scaled_data
```

## Apply the PCA Algorithm 

```python
from sklearn.decomposition import PCA
```

```python
## n_components :- it is the reduce the columns for which columns do you have to need for example 10 columns are present my dataset 
## but that is reduce the and after that reducing the columns the the 2 columns are present here then n_componets = 2
pca=PCA(n_components=2)
```

```python
pca_data=pca.fit_transform(scaled_data)
```

```python
pca.explained_variance_
```

```python
## Plot these things 
plt.figure(figsize=(8,6))
plt.scatter(pca_data[:,0],pca_data[:,1],c=cancer_data['target'],cmap='plasma')
plt.xlabel('First Principle Component ')
plt.ylabel('Second Principle Component')
```
