## Mongo Db with python - Querying Json Documents

```python
import pymongo

client=pymongo.MongoClient('mongodb://127.0.0.1:27017')

mydb=client['Employee']
empinfo=mydb.employeeinformation
```

```python
record={
    'first Name':'Madhav',
    'last Name':'Yadav',
    'Department':'Data scientist',
    'Qualification':'B.Tech',
    'Age':21
}

empinfo.insert_one(record)
```

```python
records=[
    {
    'first Name':'Subodh',
    'last Name':'Yadav',
    'Department':'Data Engineer',
    'Qualification':'B.Tech',
    'Age':29
    },
    {
    'first Name':'Suraj',
    'last Name':'Yadav',
    'Department':'Electrical Engineer',
    'Qualification':'B.Tech',
    'Age':90
    }
    
]

empinfo.insert_many(records)
```
