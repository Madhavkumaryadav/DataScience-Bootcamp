## Wordnet Lemmatize 
Lemmatization technique is like stemming. The output we will get after lemmatization is called 'lemma', which is a root word rather than root stem, the output of stemming. After lematization, we will be getting a valid word that means the same thing. 

NLTK provides WOrdNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us Understand it with an example 

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
'''
POS- Noun-n --> By default set as the NOUN
adjective -a
adverb -r
verb -v
'''
# print(WordNetLemmatizer().lemmatize('going'))
lemmatizer.lemmatize('going')
```

```python
words=['eating','eats','eaten','writing','writes','programming','programs','history']
```

```python
## Noun parameter passed like ---> pos='n' ---> It means the n->Noun 
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='n'))
```

```python
## Adjective Parameter passed like ----> pos='a' ----> It means the a->adjective 

for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='a'))
```

```python
## verb Parameter passed like ----> pos='v' ----> It means the v->verb 

for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
lemmatizer.lemmatize('goes',pos='v')
```
