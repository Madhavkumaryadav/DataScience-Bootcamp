## Named Entity Recognition 

```python
sentence = '''
The Eiffel Tower way built from 1887 to 1889 by French Engineer 
Gustave Eiffel, whose company specialized in building metal framwewords and structures.
'''

"""
Person Eg: Madhav Kumar Yadav
Place or Location Eg: India
Date Eg: September , 24-09-1889
Time Eg: 4:30 PM
Money Eg: 1 million dollar 
Organization eg : iNeuron Private Limited
Percent Eg: 20% , twenty percent 
"""
```

```python
sentence
```

```python
import nltk 
words=nltk.word_tokenize(sentence)
```

```python
tag_elements= nltk.pos_tag(words)
```

```python
nltk.download('words')
nltk.download('maxent_ne_chunker_tab')
```

```python
## Use NLTK's currently recommended named entity chunke to chunk the give list of taged tokens.
# nltk.download('maxent_ne_chunker')
# nltk.download('words')
nltk.ne_chunk(tag_elements).draw()
```
