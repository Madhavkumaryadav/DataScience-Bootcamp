## Parts of speech tagging 

### Tagging :- 
    A part-of-speech tagger, or POS-tagger, processes a sequence of words, and attaches a part of speech tag to each word (don't forget to import nltk):

1. The process of classifying words into their parts of speech and labeling them accordingly is known as part-of-speech tagging, POS-tagging, or simply tagging. Parts of speech are also known as word classes or lexical categories. 
2. The collection of tags used for a particular task is known as a tagset. Our emphasis in this chapter is on exploiting tags, and tagging text automatically

```python
import nltk 
sentences = nltk.sent_tokenize(paragraph)
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
```

```python
## We will find to POS tag 

from nltk.corpus import stopwords 

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i].lower())
    words = [word for word in words if word not in set(stopwords.words('english'))]
    # sentences[i]=' '.join(words) ## Converting all the list of words into sentences 
    pos_tag = nltk.pos_tag(words)
    print(pos_tag)
```
