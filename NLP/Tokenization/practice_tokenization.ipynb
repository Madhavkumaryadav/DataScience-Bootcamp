```python
paragraph = '''
Hello! 
My name is Madhav Kumar.
I am From Bihar.
I live in Jaipur . 
For the Study Purpos. 
'''
```

```python
for i in paragraph.split('\n'):
    print(i)
    
```

```python
## Sentence Tokenize 
import nltk 
from nltk.tokenize import sent_tokenize,word_tokenize
```

```python
sent_word=sent_tokenize(paragraph,language='english')
print(sent_word)
```

```python
# word tokenize 
for i in paragraph.split('\n'):
    
    word=word_tokenize(i,language='english')
    print(word)
word_word=word_tokenize(paragraph)
print(word_word)
```

```python
## Using Another Library 
from nltk.tokenize import wordpunct_tokenize
punct_word= wordpunct_tokenize(paragraph)
print(punct_word)
```

```python
## Using TreebankWordTokenizer
from nltk.tokenize import TreebankWordTokenizer
tree=TreebankWordTokenizer()
tree_word=tree.tokenize(paragraph)
print(tree_word)
```

```python
## Comparision of the these four tokenizer 
print("Word length of the sent_tokenize : " , len(sent_word))
print("Word length of the word_tokenize : " ,len(word_word))
print("Word Length of the wordpunct_tokenize : " , len(punct_word))
print("word Length of the TreebanWordTokenizer : " , len(tree_word))
```

## Stemming 

```python
## PorterStemmer 
words = ['eating','eat','eates','eaten','goes','go','going','went' ] 
from nltk.stem import PorterStemmer
porter_stem = PorterStemmer()
porter_word = []
for i in words:
    original=i
    stemmer=porter_stem.stem(i)
    a=[]
    a.append(original)
    a.append(stemmer)
    porter_word.append(a)
    print(i+ " ----> " + porter_stem.stem(i))
```

```python
## RegexpStemmer 
from nltk.stem import RegexpStemmer
regexp_word=[]
reg=RegexpStemmer('ing$|s$|e$|able$',min=4)
for i in words:
    original = i 
    stemming = reg.stem(i)
    a=[]
    a.append(original)
    a.append(stemming)
    regexp_word.append(a)
    print(i+"-----> " + reg.stem(i))
```

```python
len(regexp_word)==len(porter_word)
```

```python
## Snowball Stemmer 
from nltk.stem.snowball import SnowballStemmer
snowball=SnowballStemmer(language='english')
snowball_word = []
for i in words:
    original=i
    stemming=snowball.stem(i)
    a=[]
    a.append(original)
    a.append(stemming)
    snowball_word.append(a)
    print(i+ " -------> " + snowball.stem(i))
    
    
```

```python
## Compare stemming 
### PorterStemmer , regexpStemmer  and snowballstemmer 
data={
    'porter':porter_word,
    'regexp':regexp_word,
    'snowball':snowball_word
}
```

```python
import pandas as pd
d=pd.DataFrame(data)
d
```

## Lematization Text Preprocessing 

```python
# 1. WordNetLemmatizer  
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
wordnetlemmatizer=WordNetLemmatizer()
# wordnetlemmatizer.lemmatize()
wordnet_word=[]
for i in words:
    original=i
    lemmatizer = wordnetlemmatizer.lemmatize(i,pos='v')
    a=[]
    a.append(original)
    a.append(lemmatizer)
    wordnet_word.append(a)
    print(i+"------>"+wordnetlemmatizer.lemmatize(i,pos='v'))
```
