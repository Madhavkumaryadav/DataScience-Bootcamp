## Stemming 
Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma. Stemming is important in natural language Understanding (NLU) and Natural Language Processing (NLP)



```python
## Classification Problem 
## Comments of product is a positive review or negative review 
## Review ---> Eating ,eat , eaten , [going , gone, goes] ->go 
```

### Stemming Example 

```python
words=['eating','eat','eaten','writing','writes','programming','programs','history','finalize','finally']
```

#### 1. PorterStemmer

```python
from nltk.stem import PorterStemmer 
```

```python
stemming = PorterStemmer()
for word in words:
    print(word + "--->"+stemming.stem(word))
```


## RegexpStemmer class
NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basiccally takes a single regular expression and remove 
prefix or suffix that matches the expression. Let us see an example 

#### RegexpStemmer class Example 

```python
from nltk.stem import RegexpStemmer 
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

## SnowballStemmer 
Snowball stemmer is the better of the porterstemmer . porterstemmer that's getting the lot of meaning in wrong but snowball stemmer is the optimize techniques of the porterstemmer and that's also reduce the error accuracy . 


#### Some Example of the snowballStemmer 

```python
from nltk.stem import SnowballStemmer 
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"--->"+snowballstemmer.stem(word))
```
