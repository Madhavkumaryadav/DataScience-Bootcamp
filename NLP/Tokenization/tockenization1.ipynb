## NLP(Natural Language Processing) 
#### Implementation of the tokenization 

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
corpus='''Hello , My name is Madhav.
I'm passionate about data science. 
'''
```

```python
print(corpus)
```

```python
sentences=sent_tokenize(corpus)
sentences
```

## Tokenization 
### Paragraph --> Words 
### Sentence ---> WOrds 

```python
from nltk.tokenize import word_tokenize
```

```python
words=word_tokenize(corpus)
print(words)
```

```python
for sentence in sentences:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
for  i in sentences:
    print(wordpunct_tokenize(i))
```

##### Comparrision of the word_tokenize and wordpunct_tockenize

```python
wordpun=wordpunct_tokenize(corpus)
word = word_tokenize(corpus)
```

```python
print('Length of the wordpunct_tokenize is : ',len(wordpun))
print("Length of the word_tokenize is : ", len(word))
```

```python
## Using TreebankWordTokenizer() method 
from nltk.tokenize import TreebankWordTokenizer
tw=TreebankWordTokenizer()
twords=tw.tokenize(corpus)
print(twords)
```

```python
## Comparision Of Tokenizer 

print(f"Length of the TreebankWordTokenizer : {len(twords)}")
print(f"Length of the wordpunct_tokenize : {len(wordpun)}")
print("Length of the word_tokenize is : ", len(word))
```
