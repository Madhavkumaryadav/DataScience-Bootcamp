```python
import logging 
logging.basicConfig(
    filename='div_test.log',
    level=logging.INFO ,
    format = '%(asctime)s %(message)s'
)

def test(a,b):
    logging.info("This is a start of function ")
    res=a/b
    logging.info(f"Your input is {str(a)} and {str(b)}. then the divided by the result is : {res}")
    return res
```

```python
test(5,6)
```

```python
logging.shutdown()  ## then the logging is not working 
```

```python
test(4,5)
```
