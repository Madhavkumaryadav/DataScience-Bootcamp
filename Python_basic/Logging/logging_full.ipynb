```python
def test(a,b):
    try:
        lg.info(f"Your Input is numenitor : {str(a)}, and denumenitor : {str(b)}")
        div = a/b 
        lg.info(f"Your result is {str(div)}")
        return div 
    
    except Exception as e:
        print("You can check you log for more info if your code will fail ")
        lg.error("Error has occured")
        lg.exception(str(e))
```
