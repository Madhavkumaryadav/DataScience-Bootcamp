```python
# import numpy for generating random number 
import numpy as np
# import matplotlib library
import matplotlib.pyplot as plt ## plt as alias 
from matplotlib import style ## for grid styling  
%matplotlib inline     
# for displaying plot within jupyter notebook 
```

```python
## generating random number (total 10 ) 
randomno=np.random.rand(10)
print(randomno)
```

```python
## select the style  of the plot 
style.use('ggplot')
# plot the random numbers 
plt.plot(randomno,'g',label='line one',linewidth=2)
# x-axis is the number of random number (index)
plt.xlabel("Range")
# y-axis is the actual random numbers 
plt.ylabel("Numbers")
# title of the plot 
plt.title('FIrst Plot')

plt.legend() # to show what does each line represent 
plt.show() # plot the graph as output 
```
